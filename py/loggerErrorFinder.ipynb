```python
import os
from datetime import datetime

SEARCH_STRING = "org.littletonrobotics.junction.Logger.periodicAfterUser"
ALLOWED_EXTENSIONS = (".dsevents",)

def parse_datetime_from_filename(file_path):
    """Extract datetime from filename in format YYYY_MM_DD HH_MM_SS"""
    filename = os.path.basename(file_path)
    name, _ = os.path.splitext(filename)
    try:
        return datetime.strptime(name, "%Y_%m_%d %H_%M_%S")
    except ValueError:
        return datetime.min  # fallback if filename doesn't match

def scan_folder(folder_path):
    with_string = []
    without_string = []

    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if not file.lower().endswith(ALLOWED_EXTENSIONS):
                continue

            file_path = os.path.join(root, file)
            count = 0

            try:
                with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
                    for line in f:
                        count += line.count(SEARCH_STRING)

                if count > 0:
                    with_string.append((file_path, count))
                else:
                    without_string.append(file_path)

            except Exception as e:
                print(f"Could not read {file_path}: {e}")

    # Sort both lists by date/time descending
    with_string.sort(key=lambda x: parse_datetime_from_filename(x[0]), reverse=True)
    without_string.sort(key=lambda x: parse_datetime_from_filename(x), reverse=True)

    return with_string, without_string


if __name__ == "__main__":
    folder = "D:\dsLogs"

    # folder = "C:\\Users\\orenm\\code\\miscar\\robot-2026\\py\\logs"
    with_str, without_str = scan_folder(folder)

    if with_str:
        print("\nFiles containing the string (sorted by date/time):\n")
        for file_path, count in with_str:
            print(f"{file_path} ({count} matches)")
    else:
        print("No files contain the string.")

    if without_str:
        print("\nFiles without the string (sorted by date/time):\n")
        for file_path in without_str:
            print(file_path)
    else:
        print("All files contain the string.")
```
